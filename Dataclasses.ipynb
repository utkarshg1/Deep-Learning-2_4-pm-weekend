```python
from dataclasses import dataclass
```

```python
@dataclass
class Person:
  name: str
  age: int
  gender: str
  occupation: str

  def __post_init__(self):

    self.validate_name()
    self.validate_age()
    self.validate_gender()
    self.validate_occupation()

  def validate_name(self):
    if not isinstance(self.name, str):
      TypeError("Expected 'name' to be a string")

  def validate_age(self):
    if not isinstance(self.age, int):
      raise TypeError("Expected 'age' to be an integer")
    elif self.age<=0:
      raise ValueError("Age must be positive integer")

  def validate_gender(self):
    if not isinstance(self.gender, str):
      raise TypeError("Expected 'gender' to be a string")
    elif self.gender not in ["male", "female"]:
      raise ValueError("Gender must be 'male' or 'female'")

  def validate_occupation(self):
    if not isinstance(self.occupation, str):
      raise TypeError("Expected 'occupation' to be string")
```

```python
from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int
    gender: str
    occupation: str

    def __post_init__(self):
        errors = []  # Collect all validation errors

        # Validation checks
        self.validate_name(errors)
        self.validate_age(errors)
        self.validate_gender(errors)
        self.validate_occupation(errors)

        # If there are any errors, raise a ValueError with all error messages
        if errors:
            raise ValueError("Validation errors:\n" + "\n".join(errors))

    def validate_name(self, errors):
        if not isinstance(self.name, str):
            errors.append(f"Expected 'name' to be a str, got {type(self.name).__name__}")

    def validate_age(self, errors):
        if not isinstance(self.age, int):
            errors.append(f"Expected 'age' to be an int, got {type(self.age).__name__}")
        elif self.age <= 0:
            errors.append("Age must be a positive integer")

    def validate_gender(self, errors):
        if not isinstance(self.gender, str):
            errors.append(f"Expected 'gender' to be a str, got {type(self.gender).__name__}")
        elif self.gender.lower() not in {"male", "female"}:
            errors.append("Gender must be 'male' or 'female'")

    def validate_occupation(self, errors):
        if not isinstance(self.occupation, str):
            errors.append(f"Expected 'occupation' to be a str, got {type(self.occupation).__name__}")

# Example Usage
try:
    person = Person("Kalyani", -23, "F", 1234)  # This will raise multiple validation errors
except ValueError as e:
    print(e)
```
